```python
import numpy as np
import msicpe.san as san
from plotly import express as px
import scipy.signal as signal
import pandas as pd
```

```python
signal_dict = np.load('piano_G.npz') # signal_dict est un dictionnaire contenant les données contenues dans le fichier signal.npz
s = signal_dict['s.npy'] # extraction du signal utile
t = signal_dict['t.npy'] # extraction du vecteur temps associé au signal

df=pd.DataFrame({"t":t,"s":s})
figure=px.line(df,x="t",y="s",title="Courbes Signal Piano")
figure.show()

nu,S = san.trans_fourier(s,t)
S_Reelle = np.real(S)
S_Imag = np.imag(S)
spectre=np.sqrt(S_Reelle**2+S_Imag**2)**2

df_TF_R=pd.DataFrame({"nu":nu,"spectre":spectre})
figure_TF_R=px.line(df_TF_R,x="nu",y="spectre",title="Spectre Signal Piano")
figure_TF_R.show()

spectre=np.log(spectre)
df_TF_R=pd.DataFrame({"nu":nu,"spectre":spectre})
figure_TF_R=px.line(df_TF_R,x="nu",y="spectre",title="Spectre en log Signal Piano")
figure_TF_R.show()
```

Maintenant que ça marche pour un seul signal d'une voyelle, on va faire la même chose pour tous les signals de la voyelle e.

```python
def vecteur_log(voy:str,H=15) -> list : 
    """
    La fonction prendra une voyelle et renverra 5 vecteurs de dimension (H +1) contenant les valeurs d’amplitude
du logarithme (base 10) du spectre aux fréquences d’intérêt.
    """
    spectre_freq=[]
    for i in range(1,6):
        dat = np.load(voy+f'_{i}.npz') 
        s = dat['s'] 
        t = dat['t'] 

        mean_s=np.mean(s)
        s=s-mean_s

        nu,S = san.trans_fourier(s,t)
        S_Reelle = np.real(S)
        S_Imag = np.imag(S)
        spectre=np.sqrt(S_Reelle**2+S_Imag**2)**2
        nu = np.real(nu) 

        max_spectre=np.max(spectre)
        spectre=spectre/max_spectre

        freq_fond=san.detect_fondamentale(spectre,nu)

        liste_frequences=[i*freq_fond for i in range(1,H+2)]

        f,A=san.detect_pics(spectre,liste_frequences,nu)

        spectre_freq.append(np.log10(A))
    return spectre_freq
```

```python
def valeur_moy_spectre(voy:str,H=15) -> list :
     """ 
     Prend en argument la voyelle à étuider et renvoie le vecteur contenant la valeur moyenne des spectres 
     pour chaque fréquence d’intérêt.  
     """
     spectre_freq_voy=np.array(vecteur_log(voy,H=15))
     vect_moy=[]
     for j in range(H+1):
          s=0
          for i in range(5): 
               s+=spectre_freq_voy[i,j]
          vect_moy.append(s/5)
     return vect_moy

def affichage_spectre_et_moy(voy:str,H=15):
    vect_moy=valeur_moy_spectre(voy)
    spectre_freq_log=vecteur_log(voy,H=15)    
    Liste_H = [i for i in range(H+1)]    
    df=pd.DataFrame({"Harmonique":Liste_H,"spectre_0":spectre_freq_log[0],"spectre_1":spectre_freq_log[1],"spectre_2":spectre_freq_log[2],
                     "spectre_3":spectre_freq_log[3],"spectre_4":spectre_freq_log[4],"spectre_moy":vect_moy})
    figure=px.line(df,x="Harmonique",y=["spectre_0","spectre_1","spectre_2","spectre_3","spectre_4","spectre_moy"],title="Spectre en log de la voyelle"+voy)
    figure.show()
```
