```python
import numpy as np
import plotly.express as px
import pandas as pd # pour les dataframes
import sounddevice as sd # pour jouer les sons associés aux signaux
```

```python
def f(A,nu0,nue,phase,K):
    k=np.array([i for i in range(K)])
    t=k*(nue)**(-1)
    s=np.cos(2*np.pi*nu0*t+phase)
    return k,t,s
```

```python
k0,t0,s0=f(1,50,1e3,0,1000)
k1,t1,s1=f(2,100,1e3,0,1000)
k2,t2,s2=f(1,0,1e3,0,1000)
s=s0+s1+s2
D=t0[-1]
print(D,"s")
K=1000

df=pd.DataFrame({"t":t1,"s":s})
figure=px.line(df,x="t",y="s",title="Courbes Signal en fonction du temps")
figure.show()

df=pd.DataFrame({"k":k1,"s":s})
figure=px.line(df,x="k",y="s",title="Courbes Signal en fonction des indices")
figure.show()
```

```python
k,t,x=f(1,800,1e3,0,500)
sd.play(x, samplerate=1e3)

df=pd.DataFrame({"k":k,"s":x})
figure=px.line(df,x="k",y="s",title="Courbes Signal en fonction des indices")
#figure2=px.scatter(df,x="k",y="s",color="#FF0000")
figure.show()
#figure2.show()
```

En changeant la valeur de nue, le résulat n'est pas le même que dans le TD, on a pas les portes et les triangles alternés. 

Le son, en diminuant la fréquence, est plus grave.

```python

def TFD(s,N,nue):
    n=np.array([i for i in range(N)])
    f=n/N
    nu=f*nue
    S=np.fft.fft(s,N)
    return n,f,nu,S

n,f,nu,S=TFD(s,4*K,5e3)

df=pd.DataFrame({"nu":nu,"S":np.real(S)})
figure=px.line(df,x="nu",y="S",title="Courbes FFT Signal en fonction de la fréquence réelle")
figure.show()

df=pd.DataFrame({"nu":nu,"S":np.imag(S)})
figure=px.line(df,x="nu",y="S",title="Courbes FFT Signal en fonction de la fréquence imaginaire")
figure.show()

```

On remarque, en réel, des pics à la fréquence 0Hz, à la fréquence 250Hz et 500Hz. On trace maintenant le logarithme du spectre en base 10. 

```python
df=pd.DataFrame({"nu":nu,"S":np.log10(np.real(S))})
figure=px.line(df,x="nu",y="S",title="Courbes FFT en log10 Signal en fonction de la fréquence réelle")
figure.show()

df=pd.DataFrame({"nu":nu,"S":np.log(np.imag(S))})
figure=px.line(df,x="nu",y="S",title="Courbes FFT en log10 Signal en fonction de la fréquence imaginaire")
figure.show()
```

On remarque que, avec le spectre en log en base 10, l'amplitude des pics est réduite, que les pics sont espacés de la même façon qu'en réel, et qu'il y a un bruit qui apparaît d'amplitude variant entre 0 et -15.
